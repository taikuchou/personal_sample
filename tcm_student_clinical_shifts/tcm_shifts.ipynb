```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import random
```

```python
# load data
df = pd.read_csv('student_shifts.csv')
df.head()
```

```python
# transpose
df_transposed = df.T
df_transposed.columns = [f'tcol_{i}' for i in range(df_transposed.shape[1])]
df_transposed.index = [i for i in range(df_transposed.shape[0])]
df_transposed = df_transposed[[f'tcol_{i}' for i in range(58)]]
df_transposed.head()
```

```python
# set parsing scope
am_shift_index = [f'tcol_{i}' for i in range(5,34)]
pm_shift_index = [f'tcol_{i}' for i in range(38,58)]
# am_shift_index,pm_shift_index
```

```python
# student names
student_list = ['Al-Juboori, Sarah', 'Baird, Jessica', 'Fan, Qingdian (Alan)', 
                'Feng, Nan (Kenna)', 'Ho, Yi-Ping (Nicole)', 'Karimpour, Masoud', 'Luo, Ziwen (Flora)', 'Salib, Monica', 'Sekhon, Harneesh', 'Sharpe, Mia', 'Smith, Annice', 'To, Soi', 'Walser, Lauryn', 'Wang, Yuchen ', 'Wu, Shu-Hui (Kate)'] 
```

```python
date_list = []
diff_dict = {}
def get_day_shifts(df):
    date_str = df["tcol_2"]
    morning_shift = df[am_shift_index].copy()
    afternoon_shift = df[pm_shift_index].copy()
    morning_shift.dropna(inplace= True)
    morning_shift.reset_index(drop=True,inplace= True)
    afternoon_shift.dropna(inplace= True)
    afternoon_shift.reset_index(drop=True,inplace= True)
    return get_date_info(date_str), morning_shift , afternoon_shift
    
def get_different(list1,list2):
    return list(set(list1) - set(list2))

def get_date_info(date_str,date_format='%d-%b-%y'):
    date = pd.to_datetime(date_str, format=date_format)
    # day_of_week = date.day_name()
    # date.dayofweek
    return date
```

```python
def get_week(df, num, is_debug = False):
    # print(f'Week {num}')
    num = num -1 
    start = 2 + num* 6
    end = 6 + num* 6 + 1
    df_week = df[start:end]
    for i in range(start,end):
        df_day = df_week.loc[i]
        date, morning_shift, afternoon_shift = get_day_shifts(df_day)
        if date.dayofweek in [2,4]:
            am_list = morning_shift.to_list()
            pm_list = afternoon_shift.to_list()
            am_dif_list = get_different(student_list,am_list)
            pm_dif_list = get_different(student_list,pm_list)
            date_list.append(date)
            date_str = date.strftime("%Y-%m-%d")
            if len(am_dif_list) != 0 and len(am_dif_list) < 15:
                if is_debug:
                    print(f"\n{date_str}_AM_DIFF({len(am_dif_list)})\n",am_dif_list)
                diff_dict[f'{date_str}_am'] = am_dif_list
            if len(pm_dif_list) != 0 and len(pm_dif_list) < 15 and is_debug:
                if is_debug:
                    print(f"\n{date_str}_PM_DIFF({len(pm_dif_list)})\n",pm_dif_list)
                diff_dict[f'{date_str}_pm'] = pm_dif_list
            if len(am_dif_list) >= len(student_list) or len(pm_dif_list) >= len(student_list):
                date_list.remove(date)
```

```python
# Parsing 15 weeks
date_list = []
diff_dict = {}
for w in range(1,16):
    get_week(df_transposed,w)
# datetime[] to str[]
str_date_list = [ts.strftime("%Y-%m-%d") for ts in date_list]
```

```python
# initializing schedule and calculate counts
cleaner_text = "Cleaner"
receptionist_text = "Receptionist"
c_assignment_count = {s: 0 for s in student_list}
r_assignment_count = {s: 0 for s in student_list}
scheduler = {}
```

```python
def schedule_shift(key, is_rand = False):
    scheduler[key] = {}
    available_students = [s for s in student_list if s not in diff_dict.get(key, [])]
    if is_rand:
        random.shuffle(available_students)
    # pick clearner 
    cleaner = min(available_students, key=lambda s: c_assignment_count[s])
    c_assignment_count[cleaner] += 1
    scheduler[key][cleaner_text] = cleaner
    available_students.remove(cleaner)  # prevent the same clearner and receptionist
    
    # pick receptionist
    receptionist = min(available_students, key=lambda s: r_assignment_count[s])
    r_assignment_count[receptionist] += 1
    scheduler[key][receptionist_text] = receptionist
```

```python
#schedule_shifts 
c_assignment_count = {s: 0 for s in student_list}
r_assignment_count = {s: 0 for s in student_list}
scheduler = {}
rand=True
suffix = "random" if rand else "fix"
for date_str in str_date_list:
    key = f'{date_str}_am' 
    schedule_shift(key)
    key = f'{date_str}_pm' 
    schedule_shift(key,is_rand=rand)
```

```python
# validation
c = 0
error = 0
for key in scheduler:
    if key in diff_dict:
        c += 1
        duplicate_students = [s for s in scheduler[key] if s in diff_dict.get(key, [])]
        if(len(duplicate_students)!=0):
            error += 1
            print("Error:",key)
# print(error,c)
if error == 0:
    print("No error!")
else:
    print("Error occured, please recheck again!")
    # c_assignment_count = {s: 0 for s in student_list}
    # r_assignment_count = {s: 0 for s in student_list}
    # scheduler = {}
```

```python
df_scheduler = pd.DataFrame.from_dict(scheduler, orient='index')
df_scheduler.reset_index(inplace=True)
df_scheduler.rename(columns={'index': 'shift'}, inplace=True)
df_scheduler.head()
```

```python
# save to file
df_scheduler.to_csv(f"tcm_student_shifts_{suffix}.csv")
```

```python
# generate the count report and save to a csv[optional]
data = []
data.append({
        '': "",
        'Cleaner Count': "Cleaner Count",
        'Receptionist Count': "Receptionist Count"
    })
for student in student_list:
    data.append({
        '': student,
        'Cleaner Count': c_assignment_count.get(student, 0),
        'Receptionist Count': r_assignment_count.get(student, 0)
    })

df_count = pd.DataFrame(data)
df_count_t = df_count.T
df_count_t.reset_index(drop=True,inplace= True)
df_count_t.columns = ["shift"]+student_list
df_count_t.head()
```

```python
# export report csv by merging counts and schedule
for key in scheduler:
    obj = scheduler[key]
    cleaner = obj[cleaner_text]
    receptionist = obj[receptionist_text]
    # new_row_dict = {s: "" for s in student_list}
    diff_list = diff_dict.get(key,[])
    new_row_dict = {s: "Cleaner" if s == cleaner else "Receptionist" if s == receptionist else "W.S" if s in diff_list else ""
    for s in student_list}
    new_row_dict["shift"] = key
    new_row = pd.DataFrame([new_row_dict])
    df_count_t = pd.concat([df_count_t, new_row], ignore_index=False)
df_count_t.reset_index(drop=True,inplace= True)
df_count_t.to_csv(f"tcm_student_total_shifts_{suffix}.csv")
```
